```python
import pandas as pd
import itertools
```

```python
# Import data frame
data = pd.read_excel('D:/Kerjaan/Project Find Combinations/Konsul Data.xlsx')
data.head()
```

```python
# Fungsi untuk mencari pasangan angka
def find_combinations(target_sum, values):
    results = []
    for r in range(2, len(values) + 1):  # Mencari kombinasi dari 2 hingga n angka
        for combination in itertools.combinations(values, r):
            if sum(combination) == target_sum:
                results.append(combination)
    return results
```

```python
# Target jumlah
target_sum = 1576350
```

```python
# Jika ingin target jumlahnya diubah menjadi angka lain, hapus tanda # di syntax di bawah ini, jalankan, masukkan angka yang diinginkan lalu klik Enter
# target_sum = int(input("Masukkan target jumlah: "))
```

```python
# Mencari kombinasi pasangan angka
values = data["Value"].tolist()
combinations = find_combinations(target_sum, values)
```

```python
# Menampilkan hasil
if combinations:
    print(f"Pasangan angka yang jika dijumlahkan menghasilkan {target_sum}:")
    for combination in combinations:
        print(combination)
else:
    print(f"Tidak ditemukan pasangan angka yang jika dijumlahkan menghasilkan {target_sum}.")
```
